## Popcorn Hack

| # | Sentence | Answer |
|---|---|---|
| 1 | The ___ on a toggle keeps track of whether it is on or off. | checkbox |
| 2 | The ___ is the visible part of the toggle that changes color when turned on. | slider |
| 3 | The `.switch` class controls position, size, padding, and ___. | alignment |
| 4 | The ___ property makes the slider have rounded edges. | border-radius |
| 5 | The `&:before` class dictates the ___ of the toggle. | white circle |

The checkbox does all the work but is invisible (`opacity: 0; width: 0; height: 0`).
Everything you see is the `.slider` span next to it, styled through `&:checked + .slider`.


## Challenges


```python
%%html
<!-- HTML_RUNNER: Challenge 1 - Find the Bug -->
<!-- THE BUG: the <span class="toggle"> wrapper was missing.
     .slider is position:absolute with inset:0, so it sizes itself against the
     nearest positioned ancestor. .toggle is the element that is 54x31 and
     position:relative -- without it the slider stretched to fill the whole page.
     THE FIX: wrap the input + slider in <span class="toggle">. -->
<div style="padding:16px;">
  <label class="switch">
    <span class="toggle">
      <!-- .checked is a plain true/false I can read straight off the input,
           which is what makes the line below so short. -->
      <input type="checkbox" id="c1-toggle"
             onchange="document.getElementById('c1-state').textContent = this.checked ? 'on' : 'off'">
      <span class="slider green"></span>
    </span>
    <strong>Working toggle</strong>
  </label>
  <p>Status: <strong id="c1-state">off</strong></p>
</div>
```

```python
%%html
<!-- HTML_RUNNER: Challenge 2 - Customize Colors -->
<!-- The theme's checked color comes from the SASS variables $blue1 and
     $green1. Those are already compiled into the site stylesheet, so to
     change them I write a more specific rule scoped to #c2-panel. -->
<style>
  #c2-panel .purple:checked + .slider { background: #A855F7; }
  #c2-panel .pink:checked   + .slider { background: #EC4899; }
  #c2-panel .orange:checked + .slider { background: #F97316; }
</style>
<div id="c2-panel" style="display:flex; flex-direction:column; gap:12px; padding:16px;">
  <label class="switch">
    <span class="toggle"><input class="purple" type="checkbox"><span class="slider"></span></span>
    Purple &mdash; #A855F7
  </label>
  <label class="switch">
    <span class="toggle"><input class="pink" type="checkbox"><span class="slider"></span></span>
    Pink &mdash; #EC4899
  </label>
  <label class="switch">
    <span class="toggle"><input class="orange" type="checkbox"><span class="slider"></span></span>
    Orange &mdash; #F97316
  </label>
</div>
<!-- No JavaScript at all here. That surprised me: :checked is a CSS
     pseudo-class, so the browser restyles the slider entirely on its own. -->
```

```python
%%html
<!-- HTML_RUNNER: Challenge 3 - Connect to Functionality -->
<div style="padding:16px;">
  <label class="switch">
    <span class="toggle">
      <!-- I used the "hidden" property instead of style.display because it keeps
           the styling in CSS and the state as one true/false value, which lines
           up exactly with what .checked already gives me. -->
      <input type="checkbox" id="c3-toggle"
             onchange="document.getElementById('c3-box').hidden = !this.checked">
      <span class="slider green"></span>
    </span>
    <strong>Show the secret message</strong>
  </label>
  <div id="c3-box" hidden
       style="margin-top:14px; padding:14px; border-radius:10px;
              background:rgba(34,197,94,0.12); border:1px solid rgba(34,197,94,0.4);">
    You flipped the toggle, so this block appeared.
  </div>
</div>
```

## Homework


```python
%%html
<!-- HTML_RUNNER: Homework - Accessibility Settings Panel -->
<!-- Four toggles, each wired to a real feature of this panel,
     plus a counter that updates live as settings are flipped. -->
<style>
  /* Every rule is prefixed with #hw-panel so this styling stays inside
     the homework widget and never leaks onto the rest of the site. */
  #hw-panel { padding:18px; border-radius:14px; background:#f6f7f9; color:#14161a;
              transition: background .3s, color .3s, font-size .3s; }
  #hw-panel.is-dark  { background:#15181d; color:#eef1f5; }
  #hw-panel.is-large { font-size:1.2rem; }
  #hw-panel .switch  { margin-bottom:12px; }
  #hw-panel .hw-hint { opacity:.7; font-size:.85em; display:block; }
  #hw-panel .hw-details { display:none; margin-top:10px; padding:12px; border-radius:10px;
                          background:rgba(59,130,246,.12); border:1px solid rgba(59,130,246,.4); }
  #hw-panel.is-details .hw-details { display:block; }

  /* The dot pulses by default; .is-still cancels the animation.
     That is the "reduced motion" toggle doing real work. */
  #hw-panel .hw-dot { width:12px; height:12px; border-radius:50%; background:#22C55E;
                      animation: hw-pulse 1.1s ease-in-out infinite; }
  #hw-panel.is-still .hw-dot { animation: none; }
  @keyframes hw-pulse { 0%,100% { transform:scale(1); opacity:1; }
                        50%     { transform:scale(1.6); opacity:.45; } }

  #hw-panel .hw-count { display:flex; align-items:center; gap:10px; margin-top:16px;
                        padding-top:14px; border-top:1px solid rgba(128,128,128,.35); }
</style>

<div id="hw-panel">
  <h3 style="margin:0 0 4px;">Accessibility Settings</h3>
  <p class="hw-hint" style="margin:0 0 16px;">Flip a switch &mdash; each one changes this panel for real.</p>

  <!-- Every toggle runs the SAME two lines of JavaScript. The only thing that
       changes is data-feature, which names the CSS class that toggle controls.
       Writing it generically means adding a fifth toggle is one more row of
       HTML with no new JavaScript at all. -->

  <label class="switch">
    <span class="toggle">
      <input type="checkbox" id="hw-dark" data-feature="is-dark"
             onchange="document.getElementById('hw-panel').classList.toggle(this.dataset.feature, this.checked);
                       document.getElementById('hw-active').textContent =
                         document.querySelectorAll('#hw-panel input[data-feature]:checked').length">
      <span class="slider green"></span>
    </span>
    <span><strong>Dark Mode</strong><span class="hw-hint">Repaint the panel with a dark background</span></span>
  </label>

  <label class="switch">
    <span class="toggle">
      <input type="checkbox" id="hw-large" data-feature="is-large"
             onchange="document.getElementById('hw-panel').classList.toggle(this.dataset.feature, this.checked);
                       document.getElementById('hw-active').textContent =
                         document.querySelectorAll('#hw-panel input[data-feature]:checked').length">
      <span class="slider green"></span>
    </span>
    <span><strong>Large Text</strong><span class="hw-hint">Increase the text size for readability</span></span>
  </label>

  <label class="switch">
    <span class="toggle">
      <input type="checkbox" id="hw-details" data-feature="is-details"
             onchange="document.getElementById('hw-panel').classList.toggle(this.dataset.feature, this.checked);
                       document.getElementById('hw-active').textContent =
                         document.querySelectorAll('#hw-panel input[data-feature]:checked').length">
      <span class="slider green"></span>
    </span>
    <span><strong>Show Details</strong><span class="hw-hint">Reveal the extra description block</span></span>
  </label>

  <label class="switch">
    <span class="toggle">
      <input type="checkbox" id="hw-motion" data-feature="is-still"
             onchange="document.getElementById('hw-panel').classList.toggle(this.dataset.feature, this.checked);
                       document.getElementById('hw-active').textContent =
                         document.querySelectorAll('#hw-panel input[data-feature]:checked').length">
      <span class="slider green"></span>
    </span>
    <span><strong>Reduced Motion</strong><span class="hw-hint">Stop the pulsing animation on the dot</span></span>
  </label>

  <div class="hw-details">
    <strong>Extra details.</strong> This block only exists on the page when
    "Show Details" is on. Different people need different settings, so each one
    is a choice instead of something hard-coded.
  </div>

  <!-- The counter never adds or subtracts. It recounts every checked box with
       querySelectorAll(...).length, so the number can never drift out of sync
       with what the switches actually show. -->
  <div class="hw-count">
    <span class="hw-dot"></span>
    <span><strong id="hw-active">0</strong> of 4 settings active</span>
  </div>
</div>
```
