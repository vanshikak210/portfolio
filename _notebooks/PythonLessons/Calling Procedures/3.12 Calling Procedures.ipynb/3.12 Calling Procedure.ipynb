```python
def report_status(location, status):
    print("Status update from " + location + ": " + status)


# Calling report_status() two times with different values
report_status("Poway", "Clear")
report_status("Downtown", "Under Investigation")
```

```python
# Helper Procedure
def is_priority(incident_type):
    return incident_type == "Medical Emergency"

# Outer Procedure
def process_report(location, incident_type):
    priority = is_priority(incident_type)
    if priority:
        print("PRIORITY: " + incident_type + " reported in " + location)
    else:
        print("STANDARD REPORT: " + incident_type + " reported in " + location)

# Test Calls
process_report("Poway", "Medical Emergency")
process_report("Poway", "Power Outage")
```

```python
# Homework Hack

# Helper procedure
def check_priority(incident_type):
    # Returns True for high-priority emergency types
    return incident_type == "Medical Emergency" or incident_type == "Fire"

# Outer procedure
def dispatch_response(location, incident_type):
    # Call check_priority and store the Boolean result
    is_priority = check_priority(incident_type)
    
    # Print appropriate response based on priority status
    if is_priority:
        print("URGENT: Immediate unit dispatched to " + location + " for " + incident_type + "!")
    else:
        print("ROUTINE: Non-emergency report logged for " + location + ": " + incident_type + ".")

# Call dispatch_response at least three times with different locations and incident types
dispatch_response("Poway", "Medical Emergency")
dispatch_response("Downtown", "Power Outage")
dispatch_response("Rancho Bernardo", "Fire")
```
