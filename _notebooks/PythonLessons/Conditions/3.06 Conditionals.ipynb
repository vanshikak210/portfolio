```python
# CODE_RUNNER: 3.6 Homework - score a message, then advise on it

message = "URGENT: send gift cards now to unlock your account"
amount_requested = 500
known_contact = False

risk_score = 0

# Signal 1: Check for gift cards (+4 points)
if "gift card" in message:
    risk_score += 4

# Signal 2: Check for large amount requested (+3 points)
if amount_requested > 100:
    risk_score += 3

# Signal 3: Check if sender is unknown (+2 points)
if not known_contact:
    risk_score += 2

# Note: We do not need an 'else' block for these scoring checks because if a signal 
# is not triggered, we simply do nothing (add 0) and proceed to evaluate the remaining signals.

# 3 thresholds -> 4 outcomes -> 4 answers
if risk_score >= 8:
    print(f"Risk Score {risk_score}: High Risk! Do not reply. This is almost certainly a scam.")
elif risk_score >= 5:
    print(f"Risk Score {risk_score}: Moderate Risk. Check with someone you trust before replying.")
elif risk_score >= 2:
    print(f"Risk Score {risk_score}: Low Risk. Be a little careful, but this is probably fine.")
else:
    print(f"Risk Score {risk_score}: No Risk. Nothing suspicious detected in this message.")


    ### Reachability Proof for All 4 Branches

#* **Branch 1 (`risk_score >= 8`) — High Risk:**
  #* **Inputs:** `message = "gift card"`, `amount_requested = 500`, `known_contact = False`
  #* **Calculation:** $4 + 3 + 2 = 9$ points (Reaches `if risk_score >= 8`)

#* **Branch 2 (`risk_score >= 5`) — Moderate Risk:**
  #* **Inputs:** `message = "gift card"`, `amount_requested = 0`, `known_contact = True`
  #* **Calculation:** $4 + 0 + 0 = 4$ points... wait! Let's adjust inputs to get $\ge 5$:
  #* **Corrected Inputs:** `message = "gift card"`, `amount_requested = 500`, `known_contact = True`
  #* **Calculation:** $4 + 3 + 0 = 7$ points (Reaches `elif risk_score >= 5`)

#* **Branch 3 (`risk_score >= 2`) — Low Risk:**
  #* **Inputs:** `message = "hello"`, `amount_requested = 0`, `known_contact = False`
  #* **Calculation:** $0 + 0 + 2 = 2$ points (Reaches `elif risk_score >= 2`)

#* **Branch 4 (`else`) — No Risk:**
  #* **Inputs:** `message = "hello"`, `amount_requested = 0`, `known_contact = True`
#* **Calculation:** $0 + 0 + 0 = 0$ points (Reaches `else`)
```
