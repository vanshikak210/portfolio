```python
# CODE_RUNNER: Homework Hack - update, append, and print a list

#homework
level_stars = [12, 18, 10]

level_stars[1] = level_stars[1] + 5   # second level becomes 23
level_stars.append(7)                 # new fourth level

print(level_stars)                    # the collection
print(len(level_stars))               # number of levels

total_stars = 0
for stars in level_stars:
    total_stars = total_stars + stars
print(total_stars)                    # total stars
```
