```python
# CODE_RUNNER: Simple Iteration Example - SFSRC zone scan

zone_readings = [34, 42, 31, 48]
alert_cutoff = 40

for reading in zone_readings:
    if reading >= alert_cutoff:
        print(str(reading) + ": SFSRC alert")
    else:
        print(str(reading) + ": normal patrol")
```
