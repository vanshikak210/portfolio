```python
# =========================
# Homework
# =========================

bonus = 5

level2_stars = level2_stars + bonus

print("\nStored Total After Bonus:", total_stars)

total_stars = level1_stars + level2_stars + level3_stars

print("Recalculated Total:", total_stars)

# Prediction:
# The stored total will remain 40 until it is recalculated.
# The recalculated total will be 45.

# Actual output:
# Stored Total After Bonus: 40
# Recalculated Total: 45

# Test 1: Bonus of 5

level1_stars = 12
level2_stars = 18
level3_stars = 10

total_stars = level1_stars + level2_stars + level3_stars

print("\nTest 1")
print("Original/Stored Total:", total_stars)

level2_stars = level2_stars + 5

print("Stored Total After Bonus:", total_stars)

total_stars = level1_stars + level2_stars + level3_stars

print("Updated Total:", total_stars)

# Prediction:
# The stored total will stay at 40 and the updated total will be 45.

# Actual output:
# Original/Stored Total: 40
# Stored Total After Bonus: 40
# Updated Total: 45


# Test 2: Bonus of 0

level1_stars = 12
level2_stars = 18
level3_stars = 10

bonus = 0

level2_stars = level2_stars + bonus

total_stars = level1_stars + level2_stars + level3_stars

print("\nTest 2")
print("Total with 0 Bonus:", total_stars)

# Prediction:
# The total will remain 40 because the bonus is 0.

# Actual output:
# Total with 0 Bonus: 40
```
